# Notebook: VQE Live (simplified)

Este notebook demonstra execução local da pipeline VQE (fallback por diagonalização) e mostra como alternar para Qiskit se disponível.


```python
import numpy as np
from core.hamiltonian_builder import build_hamiltonian
from core.vqe_formation_optimizer import VQEFormationOptimizer
import matplotlib.pyplot as plt

n = 3
H = build_hamiltonian(n_qubits=n)
vqe = VQEFormationOptimizer()
energy, params, runtime = vqe.run_vqe(H)
print(f"Energy: {energy:.6f}, runtime={runtime:.3f}s")

# plot placeholder: eigenvalue spectrum
w,_ = np.linalg.eigh(H)
plt.plot(np.sort(w), marker='o')
plt.title('Eigenvalue spectrum (Hamiltonian)')
plt.xlabel('Index')
plt.ylabel('Eigenvalue')
plt.show()
```

Observações:
- Se Qiskit estiver instalado, modifique o VQEFormationOptimizer para usar VQE com RealAmplitudes e COBYLA.
- Este notebook é propositalmente pequeno para garantir execução rápida em CI/local.
